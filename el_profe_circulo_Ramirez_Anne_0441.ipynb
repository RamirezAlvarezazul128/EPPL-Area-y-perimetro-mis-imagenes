# Formula para calcular Area

---
# Formula para calcular perimetro

```python
# Calcular área y el perímetro de un círculo
# variables A,P,r
# formulas A = pi * r^2 y P = 2 * pi * r
# Capturar el valor de r
import math
print("Calcular área y el perímetro de un círculo")
r = float(input("Ingrese el valor del radio: "))
A = math.pi * r**2
P = 2 * math.pi * r

print("El área del círculo es: ", A)
print("El perímetro del círculo es: ", P)
print("Ramirez Anne NL = 44")
```
