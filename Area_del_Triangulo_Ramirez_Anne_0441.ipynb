# Calcular Area de un triangulo
---
![Fórmulas del area](https://www.neurochispas.com/wp-content/uploads/2021/03/formula-del-area-de-un-triangulo-rectangulo.png)

---
# Formula para calcular Area
$$A = \frac{b \cdot h}{2}$$
---
## Datos de entrada
* $b$ = ?
* $h$ = ?
---
# Diagrama de flujo en LucidChart
![Fórmulas del area](https://raw.githubusercontent.com/RamirezAlvarezazul128/EPPL-Area-y-perimetro-mis-imagenes/refs/heads/main/DF%20Area%20del%20triangulo.png)



```python
# Calcular el área de un triángulo
# Variables: b (base), h (altura), A (área)
# Fórmula: A = (base * altura) / 2
# Capturar los valores de b y h

print("Calcular el área de un triángulo")

b = int(input("Ingrese el valor de la base: "))
h = int(input("Ingrese el valor de la altura: "))

# Aplicamos la fórmula A = (b * h) / 2
A = (b * h) / 2

print("El área del triángulo es: ", A)
print("Programa realizado por Ramirez Anne NL = 44")
```
